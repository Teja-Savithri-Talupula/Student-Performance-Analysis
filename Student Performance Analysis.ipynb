```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\talup\\OneDrive\\Documents\\student_data.csv")
print(df)
```

```python
df['Average'] = df[['Stats','Python','Excel','SQL','Power BI']].mean(axis=1)
print(df['Average'])
```

```python
plt.figure(figsize=(15, 7))
plt.bar(df['Name'], df['Average'], color='skyblue')
plt.title('Average Scores of Students')
plt.xlabel('Student')
plt.ylabel('Average Score')
plt.show()
```
